```python
fig= px.scatter(df, x='YearsExperience',y='Salary',title='Experience vs Salary')
fig.show()
```

```python
x=df["YearsExperience"]
y=df["Salary"]
print(type(x))
```

```python
x_mean=x.mean()
y_mean=y.mean()
print(x_mean)
print(y_mean)
```

```python
# m-->slope 
# b--->intercept 

# m= summation i=1->n (xi - x_mean )(yi - y_mean) / summation i=1->n (xi - x_mean)^2

# b= y_mean - m*x_mean

# total_error = summation i=1->n (yi-mxi-b)^2
numerator = 0
denominator = 0

for index, row in df.iterrows():
    numerator += (row["YearsExperience"] - x_mean) * (row["Salary"] - y_mean)

    denominator += (row["YearsExperience"] - x_mean) ** 2

m = numerator / denominator

b = y_mean - m * x_mean

total_error = 0

for index, row in df.iterrows():
    y_pred = m * row["YearsExperience"] + b

    total_error += (row["Salary"] - y_pred) ** 2

print("Slope m =", m)
print("Intercept b =", b)
print("Total Error =", total_error)

    
```

```python
df["y_pred"] = m * x + b

fig = px.scatter(df, x="YearsExperience", y="Salary")

fig.add_traces(
    px.line(
        df.sort_values("YearsExperience"),
        x="YearsExperience",
        y="y_pred"
    ).data
)

fig.show()
```
